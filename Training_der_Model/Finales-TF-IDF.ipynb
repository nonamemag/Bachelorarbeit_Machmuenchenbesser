```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV
import joblib

# 1. Daten einlesen
trainingsdaten = pd.read_json("train_TFIDF.json")
merkmale_X = trainingsdaten['text']
zielvariable_y = trainingsdaten['service_name']

# 2. Vektorisierung
wort_zaehler = TfidfVectorizer()
merkmale_X_zahlen = wort_zaehler.fit_transform(merkmale_X)

# 3. Cross-Validation Setup
param_grid = {'C': [0.01, 0.1, 1.0, 10.0, 100.0]}
basis_modell = LogisticRegression(max_iter=1000, random_state=42)

suche = GridSearchCV(basis_modell, param_grid, cv=5, scoring='f1_macro')

print("Starte Cross-Validation für TF-IDF...")
suche.fit(merkmale_X_zahlen, zielvariable_y)

print(f"Bester Parameter für TF-IDF: {suche.best_params_}")

# 4. Persistierung mittels joblib
joblib.dump(suche.best_estimator_, 'trainiertes_tfidf_modell.joblib')
joblib.dump(wort_zaehler, 'tfidf_vectorizer.joblib')
print("Trainiertes TF-IDF-Modell gespeichert.")
```
